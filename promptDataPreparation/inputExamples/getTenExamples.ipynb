```python
import pandas as pd
import numpy as np
import random
import torch
import os
import csv
```

```python
# Random States
def set_random_states(random_state):
    # Set various random seeds.
    np.random.seed(random_state)
    random.seed(random_state)
    torch.manual_seed(random_state)
    torch.cuda.manual_seed_all(random_state)
    os.environ["PYTHONHASHSEED"] = str(random_state)
    os.environ["TOKENIZERS_PARALLELISM"] = "false"
    try:
        torch.use_deterministic_algorithms(True)
    except Exception:
        pass
    return random_state
RANDOM_STATE = set_random_states(1618)
```

```python
dataframe_examples_dict = {}
parent_path = '../../getText/datasetsPrep'
for folder in os.listdir(f'{parent_path}'):
    if os.path.isdir(f'{parent_path}/{folder}'):
        for file in os.listdir(f'{parent_path}/{folder}'):
            if file.endswith('.csv'):
                df = pd.read_csv(f'{parent_path}/{folder}/{file}')
                df = df.sample(frac=1, random_state=RANDOM_STATE).dropna()
                dataframe_examples_dict[file.replace('.csv', '')] = df.head(10)['text'].values.tolist()
pd.DataFrame(dataframe_examples_dict).to_csv('./oneShotInputs.csv', index=False, quoting=csv.QUOTE_ALL, encoding='utf-8', lineterminator='\n')
```
