# Equal Weights for All Metrics

```python
# There should be a minimum standard deviation, otherwise the metric is returning all 0s or exhibits no variation (which we assume is not indicative of a perfect metric, but a failing one).
min_std = 1e-6
# Get standard deviation. 
std_dev = real_df.std(ddof=1)
# Get mean. 
mean = real_df.mean()

weights = ((1 - (mean * 0)) / (1 + (mean * 0)).sum())
weights

# Calculate framework score.
distance = all_metric_comparisons.copy()
distance["Framework Score"] = (
    distance[metric_cols] * weights
).sum(axis=1)

merged = all_human_comparisons.merge(distance, on="model", how="inner")
weight_rank = weights.rank(ascending=False, method="dense").astype(int)
merged = merged.merge(all_individual_metrics, on='model', how="inner")

merged = merged.rename(columns={
    "final_score": "Framework Score",
    "bertscore": "BERTScore",
    "meteor": "METEOR",
    "abs_sentiment_diff": "Abs Sentiment Diff",
    "abs_subjectivity_diff": "Abs Subjectivity Diff",
    "original_metric_mean": "Original Metric Mean",
})

merged[['model', 'F1 Score', "Cohen's Kappa", 'Framework Score', 'Original Metric Mean']]

all_results_for_df = []
for comparison in ["F1 Score", "Cohen's Kappa"]:
    for metric in ["CHI", "ZIPF", "CLASSIFIER", "IRPR", "FID", "PR", "DC", "MAUVE", "TRADITIONAL", "ZERO", "Framework Score", "BERTScore", "METEOR", "Abs Sentiment Diff", "Abs Subjectivity Diff", "Original Metric Mean"]:
        x = merged[metric]
        y = merged[comparison]


        # Pearson is standardized and works best for linear relationships. 
        result = permutation_test(
            (x, y),
            lambda x, y: pearsonr(x, y).statistic,
            permutation_type="pairings",
            alternative="two-sided",
            n_resamples=np.inf
        )
        p_corr, pp_value = result.statistic, result.pvalue

        # Spearman's does not need linearity, robust to outliers, no distributional assumptions (non normal distribution), versatile in variable types.
        result = permutation_test(
            (x, y),
            lambda x, y: spearmanr(x, y).statistic,
            permutation_type="pairings",
            alternative="two-sided",
            n_resamples=np.inf
        )
        s_corr, sp_value = result.statistic, result.pvalue

        # Kendall's Tau as there is a very small sample size. 
        result = permutation_test(
            (x, y),
            lambda x, y: kendalltau(x, y).statistic,
            permutation_type="pairings",
            alternative="two-sided",
            n_resamples=np.inf
        )
        tau, kp_value = result.statistic, result.pvalue


        all_results_for_df.append({
        'metric': metric,
        'comparison': comparison,
        'pearson': p_corr,
        'pearson p-value': pp_value,
        'spearman': s_corr,
        'spearman p-value': sp_value,
        'kendall': tau, 
        'kendall p-value': kp_value,
        })

    plt.scatter(
    merged["Framework Score"],
    merged[comparison]
    )

    plt.xlabel("Final Calibrated Score")
    plt.ylabel(comparison)
    plt.title(f"Correlation: Final Calibrated Score vs {comparison}")
    plt.show()

    plt.scatter(
    merged["Original Metric Mean"],
    merged[comparison]
    )

    plt.xlabel("Final Individual Overall Score")
    plt.ylabel(comparison)
    plt.title(f"Correlation: Final Individual Overall Score vs {comparison}")
    plt.show()

statistical_tests = pd.DataFrame(all_results_for_df)
statistical_tests.to_csv(f'./ablation_ValidityNone_ReliabilityNone.csv')
print(statistical_tests[statistical_tests['metric'] == 'Framework Score'])
print(statistical_tests[statistical_tests['metric'] == 'Original Metric Mean'])
```

# Using Weighted [Reciprocal of Coefficient of Variation](https://doi.org/10.1016/B978-0-12-381375-6.00029-2) and Validity Weights 

```python
for val_ratio, rel_ratio in zip([round(0.1 * i, 1) for i in range(0, 11)], [round(0.1 * i, 1) for i in range(10, -1, -1)]):
    # There should be a minimum standard deviation, otherwise the metric is returning all 0s or exhibits no variation (which we assume is not indicative of a perfect metric, but a failing one).
    min_std = 1e-6
    # Get standard deviation. 
    std_dev = real_df.std(ddof=1)
    # Get mean. 
    mean = real_df.mean()

    # Remove saturated metrics
    valid_metrics = pd.Series(True, index=metric_cols)
    for m in metric_cols:
        # Real-vs-real values for this metric
        real_values = real_df[m]
        # Detect saturation at minimum or maximum or standard deviation is less than min_std
        if (np.allclose(real_values, 1, atol=min_std)
            or np.allclose(real_values, 0, atol=min_std)
            or std_dev[m] < min_std
        ):
            valid_metrics[m] = False

    # Get validity weights (as inverse score such that lower scores get higher weight) and remove saturated metrics. 
    validity_weights  = (1 - mean).where(valid_metrics, 0)
    # Normalize (relative importance, not arbitrary magnitude). 
    validity_weights = validity_weights / validity_weights.sum()

    # Get reliability weights (coefficient of variation, which is dimensionless and scale invariant).
    safe_std = std_dev.clip(lower=min_std)
    reliability_weights = 1 / (1 + (safe_std / mean))
    # Remove saturated metrics. 
    reliability_weights = reliability_weights.where(valid_metrics, 0)
    # Normalize (relative importance, not arbitrary magnitude). 
    reliability_weights = reliability_weights / reliability_weights.sum()

    # Combine weights. 
    weights = ((validity_weights * val_ratio) + (reliability_weights * rel_ratio)) / 2
    # Normalize (relative importance, not arbitrary magnitude). 
    weights = weights / weights.sum()

    # Calculate framework score.
    distance = all_metric_comparisons.copy()
    distance["Framework Score"] = (
        distance[metric_cols] * weights
    ).sum(axis=1)

    merged = all_human_comparisons.merge(distance, on="model", how="inner")
    weight_rank = weights.rank(ascending=False, method="dense").astype(int)
    merged = merged.merge(all_individual_metrics, on='model', how="inner")

    merged = merged.rename(columns={
        "final_score": "Framework Score",
        "bertscore": "BERTScore",
        "meteor": "METEOR",
        "abs_sentiment_diff": "Abs Sentiment Diff",
        "abs_subjectivity_diff": "Abs Subjectivity Diff",
        "original_metric_mean": "Original Metric Mean",
    })

    merged[['model', 'F1 Score', "Cohen's Kappa", 'Framework Score', 'Original Metric Mean']]

    all_results_for_df = []
    for comparison in ["F1 Score", "Cohen's Kappa"]:
        for metric in ["CHI", "ZIPF", "CLASSIFIER", "IRPR", "FID", "PR", "DC", "MAUVE", "TRADITIONAL", "ZERO", "Framework Score", "BERTScore", "METEOR", "Abs Sentiment Diff", "Abs Subjectivity Diff", "Original Metric Mean"]:
            x = merged[metric]
            y = merged[comparison]


            # Pearson is standardized and works best for linear relationships. 
            result = permutation_test(
                (x, y),
                lambda x, y: pearsonr(x, y).statistic,
                permutation_type="pairings",
                alternative="two-sided",
                n_resamples=np.inf
            )
            p_corr, pp_value = result.statistic, result.pvalue

            # Spearman's does not need linearity, robust to outliers, no distributional assumptions (non normal distribution), versatile in variable types.
            result = permutation_test(
                (x, y),
                lambda x, y: spearmanr(x, y).statistic,
                permutation_type="pairings",
                alternative="two-sided",
                n_resamples=np.inf
            )
            s_corr, sp_value = result.statistic, result.pvalue

            # Kendall's Tau as there is a very small sample size. 
            result = permutation_test(
                (x, y),
                lambda x, y: kendalltau(x, y).statistic,
                permutation_type="pairings",
                alternative="two-sided",
                n_resamples=np.inf
            )
            tau, kp_value = result.statistic, result.pvalue


            all_results_for_df.append({
            'metric': metric,
            'comparison': comparison,
            'pearson': p_corr,
            'pearson p-value': pp_value,
            'spearman': s_corr,
            'spearman p-value': sp_value,
            'kendall': tau, 
            'kendall p-value': kp_value,
            })

    statistical_tests = pd.DataFrame(all_results_for_df)
    statistical_tests.to_csv(f'./ablation_Validity{val_ratio}_Reliability{rel_ratio}.csv')
    print(f"---------- Validity {val_ratio} vs Reliability {rel_ratio} ----------")
    print(statistical_tests[statistical_tests['metric'] == 'Framework Score'])
    print(statistical_tests[statistical_tests['metric'] == 'Original Metric Mean'])
```

```python
all_results = []

for file in os.listdir("./"):

    if not os.path.isdir(f"./{file}"):

        if file.endswith(".csv") and "ablation" in file:

            match = re.search(
                r"Validity(\d+(?:\.\d+)?)_Reliability(\d+(?:\.\d+)?)\.csv$",
                file
            )

            if match is not None:

                validity = float(match.group(1))
                reliability = float(match.group(2))
            elif file == "ablation_ValidityNone_ReliabilityNone.csv":

                validity = None
                reliability = None

            else:
                print(f"Could not parse filename: {file}")
                continue


            temp_df = pd.read_csv(f"./{file}")

            temp_df = temp_df[
                temp_df["metric"] == "Framework Score"
            ].drop(
                columns=["metric", "Unnamed: 0"],
                errors="ignore"
            )

            temp_df["validity"] = validity
            temp_df["reliability"] = reliability
            temp_df["filename"] = file

            all_results.append(temp_df)


results = pd.concat(
    all_results,
    ignore_index=True
)
```

```python
df_flat = (
    results.pivot(
        index=["validity", "reliability", "filename"],
        columns="comparison",
        values=["pearson", "pearson p-value",
                "spearman", "spearman p-value",
                "kendall", "kendall p-value"]
    )
    .reset_index()
)

df_flat.columns = [
    "_".join(col).strip("_") if isinstance(col, tuple) else col
    for col in df_flat.columns
]

df_flat.drop(columns=[column for column in df_flat.columns if 'spearman' in column or 'kendall' in column or 'filename' in column])
```

```python
y_lower = -1.0
y_upper = -0.98

fig, ax = plt.subplots(figsize=(10, 5))


line_colors = {}

for comparison in ["F1 Score", "Cohen's Kappa"]:

    subset = results[
        results["comparison"] == comparison
    ].copy()

    normal = subset[
        subset["validity"].notna()
    ].sort_values(
        "validity",
        ascending=False
    )

    equal_weights = subset[
        subset["validity"].isna()
    ]

    subset = pd.concat(
        [normal, equal_weights],
        ignore_index=True
    )

    x = np.arange(len(subset))

    y = subset["pearson"].values
    p = subset["pearson p-value"].values

    line, = ax.plot(
        x,
        y,
        linewidth=2,
        label=comparison
    )

    # Store the colour used for this line
    line_colors[comparison] = line.get_color()

    significant = p < 0.05

    ax.scatter(
        x[significant],
        y[significant],
        s=60,
        color=line.get_color(),
        zorder=3
    )

    ax.scatter(
        x[~significant],
        y[~significant],
        s=60,
        facecolors="none",
        edgecolors=line.get_color(),
        linewidths=1.5,
        zorder=3
    )


normal = results[
    (results["comparison"] == "F1 Score") &
    (results["validity"].notna())
].sort_values(
    "validity",
    ascending=False
)

equal_weights = results[
    (results["comparison"] == "F1 Score") &
    (results["validity"].isna())
]

n_normal = len(normal)

labels = [
    f"{v:.1f}"
    for v in normal["validity"]
]

if len(equal_weights) > 0:
    labels.append("Equal\nweights")

ax.set_xticks(
    np.arange(len(labels))
)

ax.set_xticklabels(labels)

ax.set_xlabel(
    "Validity → Reliability"
)

ax.set_ylabel(
    "Pearson correlation"
)

if len(equal_weights) > 0:

    equal_x = n_normal

    ax.axvline(
        equal_x - 0.5,
        linestyle=":",
        linewidth=1,
        alpha=0.7
    )

ax.set_ylim(
    y_lower,
    y_upper
)


ax.axhline(
    0,
    linestyle="--",
    linewidth=1,
    alpha=0.5
)


ax.grid(
    axis="y",
    alpha=0.2
)


line_handles = [
    Line2D(
        [0],
        [0],
        color=line_colors["F1 Score"],
        linewidth=2,
        label="F1 Score"
    ),
    Line2D(
        [0],
        [0],
        color=line_colors["Cohen's Kappa"],
        linewidth=2,
        label="Cohen's Kappa"
    )
]

sig_handles = [
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="None",
        markersize=7,
        markerfacecolor="black",
        markeredgecolor="black",
        label="p < .05"
    ),
    Line2D(
        [0],
        [0],
        marker="o",
        markerfacecolor="none",
        markeredgecolor="black",
        linestyle="None",
        markersize=7,
        label="p ≥ .05"
    )
]

ax.legend(
    handles=line_handles + sig_handles,
    frameon=False
)


plt.tight_layout()
plt.savefig('./AblationTests.png', dpi='figure', bbox_inches=None)
plt.show()
```
