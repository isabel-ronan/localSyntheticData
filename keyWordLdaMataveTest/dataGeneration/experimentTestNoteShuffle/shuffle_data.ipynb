```python
# import required libraries
import pandas as pd
import polars as pl
import os
```

```python
# load all files from data folder (generated in Google Colab)
dfs = {}
for folder in os.listdir('../processedData/'):
    for file in os.listdir(f"../processedData/"):
        if '.csv' in file:
            temp = pd.read_csv(f'../processedData/{file}')
            dfs[file.replace('.csv', '')] = temp
```

```python
dfs['LDA']
```

```python
for topic_model, df in dfs.items():
    # print(topic_model)
    print(f"{topic_model}")
    print(f"Number of Total Notes: {len(df)}")
    shuffled_df = df.sample(frac=1, random_state=64)
    shuffled_df = shuffled_df.reset_index(drop=True)
    sample_size = round(len(shuffled_df) / 10)
    # isolate the first 10% of notes
    temp_dfs = []
    for needs in ['met', 'unmet']:
        temp_dfs.append(shuffled_df[shuffled_df['needs'] == needs].iloc[:round(sample_size / 2)])
    send_df = pd.concat(temp_dfs)
    print(f"Number of Sampled Notes: {len(send_df)}")
    # print(send_df)
    # output one file for own reference
    send_df.to_excel(f'./{topic_model}_labelled_notes.xlsx', index=False)
    # output one file to send for review
    df_to_send = send_df['report']
    df_to_send.to_excel(f'./{topic_model}_unlabelled_notes.xlsx', index = False)
    for folder in os.listdir(f'./sent'):
        # Save each file to folder under fake name (MATAVE == file1 and LDA == file2)
        if topic_model == 'MATAVE':
            df_to_send.to_excel(f'./sent/{folder}/file1UnlabelledNotes.xlsx', index = False)
        elif topic_model == 'LDA':
            df_to_send.to_excel(f'./sent/{folder}/file2UnlabelledNotes.xlsx', index = False)
        else:
            print("Something went wrong!")
```
