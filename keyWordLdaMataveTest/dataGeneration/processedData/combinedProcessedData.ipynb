```python
import pandas as pd
import os
```

```python
temp_dfs = []
for file in os.listdir('./'):
    if file.endswith('.csv'):
        temp_dfs.append(pd.read_csv(f'./{file}'))
```

```python
((pd.concat(temp_dfs).sample(frac=1, random_state=1618)).reset_index()).to_csv(f'./allTopicModels.csv')
```
